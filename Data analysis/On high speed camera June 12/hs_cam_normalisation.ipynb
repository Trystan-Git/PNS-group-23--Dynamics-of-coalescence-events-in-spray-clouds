```python
import numpy as np
import matplotlib.pyplot as plt 
from lmfit import models
```

```python
average_velocity = [4.66, 11.09, 17.41]
flow_rate = [1, 2, 3]
standard_deviation_velocity = [0.22, 0.40, 0.41]
ratio = [4.66/0.22, 11.09/0.4, 17.41/0.41]
```

```python
plt.figure()
plt.errorbar(flow_rate, ratio, fmt='.', color="steelblue")

plt.title(r"Ratio between average velocity and standard deviation per flow rate")
plt.xlabel(r"Flow rate (ml/min)")
plt.ylabel(r"Ratio ($\overline{v} / \sigma_v$)")
plt.grid(True)
```

```python
def fit_function(x, a, b):
    ratio_fit = a * x + b 
    return ratio_fit
```

```python
fit_model = models.Model(fit_function)
fit_result = fit_model.fit(fit_function(ratio_fit), x=ratio, a = 1, b = 0)
```
